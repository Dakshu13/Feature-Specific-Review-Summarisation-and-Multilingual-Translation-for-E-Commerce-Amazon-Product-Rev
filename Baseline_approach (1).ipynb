```python
# CELL 11: Merge and filter (same as main notebook)
merged_df = pd.merge(a, b, on='parent_asin', how='left')
merged_df.rename(columns={
    'title_x': 'review_title', 'title_y': 'product_title',
    'images_x': 'review_images', 'images_y': 'product_images'
}, inplace=True)
merged_df.drop(columns=['price', 'bought_together'], inplace=True)
merged_df = merged_df[merged_df['helpful_vote'] > 2]
print(f"Filtered dataset: {merged_df.shape}")
```

```python
# CELL 12 FINAL: TF-IDF Extractive Summarisation
from sklearn.feature_extraction.text import TfidfVectorizer
from nltk.tokenize import sent_tokenize
import numpy as np
import re

def clean_text(text):
    text = re.sub(r'<.*?>', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

feature_keywords = {
    "Product Quality"          : ["quality", "product", "item", "material", "cheap", "broke", "durable"],
    "Value for Money"          : ["price", "value", "worth", "expensive", "cost", "money", "overpriced"],
    "Packaging & Presentation" : ["packaging", "package", "box", "wrapped", "presentation", "damaged", "sealed"],
    "Delivery & Shipping"      : ["delivery", "shipping", "shipped", "arrived", "late", "fast", "delay"],
    "Variety & Curation"       : ["variety", "selection", "curated", "choice", "different", "unique", "range"],
    "Customer Service"         : ["service", "support", "refund", "response", "helpful", "rude", "resolved"]
}

def tfidf_summarise_sentences(reviews, top_n=2):
    all_sentences = []
    for review in reviews:
        review = clean_text(review)
        try:
            sents = sent_tokenize(review)
            sents = [s.strip() for s in sents
                     if 10 <= len(s.split()) <= 50]  # max 50 words
            all_sentences.extend(sents)
        except:
            continue

    if not all_sentences:
        return "No summary available."

    vectorizer = TfidfVectorizer(stop_words='english', max_features=500)
    try:
        tfidf_matrix = vectorizer.fit_transform(all_sentences)
    except:
        return all_sentences[0]

    scores = np.array(tfidf_matrix.sum(axis=1)).flatten()
    top_indices = np.argsort(scores)[::-1]

    seen, selected = set(), []
    for idx in top_indices:
        sent = all_sentences[idx]
        if sent not in seen:
            seen.add(sent)
            selected.append((idx, sent))
        if len(selected) == top_n:
            break

    selected.sort(key=lambda x: x[0])
    return " ".join([s for _, s in selected])

def tfidf_feature_summarise_sentences(reviews, keywords, top_n=1):
    all_sentences = []
    for review in reviews:
        review = clean_text(review)
        try:
            sents = sent_tokenize(review)
            sents = [s.strip() for s in sents
                     if 10 <= len(s.split()) <= 50]  # max 50 words
            all_sentences.extend(sents)
        except:
            continue

    relevant = [s for s in all_sentences
                if any(kw in s.lower() for kw in keywords)]

    if not relevant:
        relevant = all_sentences

    if len(relevant) < 2:
        return relevant[0] if relevant else "No summary available."

    vectorizer = TfidfVectorizer(stop_words='english', max_features=300)
    try:
        tfidf_matrix = vectorizer.fit_transform(relevant)
        scores = np.array(tfidf_matrix.sum(axis=1)).flatten()
        top_indices = np.argsort(scores)[::-1]
        seen, selected = set(), []
        for idx in top_indices:
            sent = relevant[idx]
            if sent not in seen:
                seen.add(sent)
                selected.append(sent)
            if len(selected) == top_n:
                break
        return " ".join(selected)
    except:
        return relevant[0]

# Run
all_reviews = merged_df['text'].dropna().tolist()
print(f"Total reviews: {len(all_reviews)}")

print("Generating TF-IDF Overall Summary...")
tfidf_overall = tfidf_summarise_sentences(all_reviews, top_n=2)

print("Generating TF-IDF Feature Summaries...")
tfidf_feature_summaries = {}
for feature, keywords in feature_keywords.items():
    tfidf_feature_summaries[feature] = tfidf_feature_summarise_sentences(
        all_reviews, keywords, top_n=1)
    print(f"  ✅ {feature}")

# Build summary dict
tfidf_summary_clean = {
    "Overall Summary": {
        "Summary"   : tfidf_overall,
        "Key Themes": "Quality, Value for Money, Customer Service, Variety"
    },
    "Feature-Based Summary": {
        feat: {"Summary": summ}
        for feat, summ in tfidf_feature_summaries.items()
    }
}

# Print
print("\n" + "="*65)
print("     TF-IDF EXTRACTIVE SUMMARY (FINAL)")
print("="*65)
print(f"\n📌 Overall Summary:\n{tfidf_overall}\n")
for feat, summ in tfidf_feature_summaries.items():
    print(f"📌 {feat}:\n{summ}\n")
```

```python
# Step 2: Translate TF-IDF summary using Google Translate
from deep_translator import GoogleTranslator

def translate_tfidf_summary(summary, target_lang):
    translated = {
        'Overall Summary': {
            'Summary'   : GoogleTranslator(source='en', target=target_lang).translate(summary['Overall Summary']['Summary']),
            'Key Themes': GoogleTranslator(source='en', target=target_lang).translate(summary['Overall Summary']['Key Themes'])
        },
        'Feature-Based Summary': {
            feat: {'Summary': GoogleTranslator(source='en', target=target_lang).translate(details['Summary'])}
            for feat, details in summary['Feature-Based Summary'].items()
        }
    }
    return translated

# Translate
tfidf_tamil_summary     = translate_tfidf_summary(tfidf_summary_clean, 'ta')
tfidf_malayalam_summary = translate_tfidf_summary(tfidf_summary_clean, 'ml')

# Print Tamil
print("="*60)
print("     TF-IDF SUMMARY — TAMIL (Google Translate)")
print("="*60)
print(f"\n📌 Overall Summary:\n{tfidf_tamil_summary['Overall Summary']['Summary']}\n")
for feat, details in tfidf_tamil_summary['Feature-Based Summary'].items():
    print(f"📌 {feat}:\n{details['Summary']}\n")

# Print Malayalam
print("="*60)
print("     TF-IDF SUMMARY — MALAYALAM (Google Translate)")
print("="*60)
print(f"\n📌 Overall Summary:\n{tfidf_malayalam_summary['Overall Summary']['Summary']}\n")
for feat, details in tfidf_malayalam_summary['Feature-Based Summary'].items():
    print(f"📌 {feat}:\n{details['Summary']}\n")
```

```python
# CELL: Evaluate TF-IDF Google Translate — Tamil and Malayalam
from sacrebleu.metrics import CHRF
from rouge_score import rouge_scorer
from langdetect import detect
from deep_translator import GoogleTranslator
import pandas as pd

chrf   = CHRF()
scorer = rouge_scorer.RougeScorer(['rougeL'], use_stemmer=True)

def back_translate(text, src):
    try:
        return GoogleTranslator(source=src, target='en').translate(text)
    except:
        return ""

def evaluate_tfidf(tfidf_clean, tfidf_translated, src_lang, lang_name):
    results = []

    # Overall Summary
    orig  = tfidf_clean['Overall Summary']['Summary']
    trans = tfidf_translated['Overall Summary']['Summary']
    back  = back_translate(trans, src_lang)
    orig_words = set(orig.lower().split())
    back_words = set(back.lower().split())
    results.append({
        'Feature'        : 'Overall Summary',
        'chrF Score'     : round(chrf.sentence_score(back, [orig]).score, 2),
        'ROUGE-L'        : round(scorer.score(orig, back)['rougeL'].fmeasure, 3),
        'Word Overlap %' : round(len(orig_words & back_words) / len(orig_words) * 100, 1),
        'Lang Detected'  : detect(back)
    })

    # Feature summaries
    orig_feats  = tfidf_clean['Feature-Based Summary']
    trans_feats = tfidf_translated['Feature-Based Summary']
    for feat in orig_feats:
        orig  = orig_feats[feat]['Summary']
        trans = trans_feats[feat]['Summary']
        back  = back_translate(trans, src_lang)
        if not back.strip():
            continue
        orig_words = set(orig.lower().split())
        back_words = set(back.lower().split())
        results.append({
            'Feature'        : feat,
            'chrF Score'     : round(chrf.sentence_score(back, [orig]).score, 2),
            'ROUGE-L'        : round(scorer.score(orig, back)['rougeL'].fmeasure, 3),
            'Word Overlap %' : round(len(orig_words & back_words) / len(orig_words) * 100, 1),
            'Lang Detected'  : detect(back)
        })

    df = pd.DataFrame(results)
    print(f"\n{'='*65}")
    print(f"  TF-IDF — {lang_name.upper()} EVALUATION (Google Translate)")
    print(f"{'='*65}")
    print(df[['Feature','chrF Score','ROUGE-L','Word Overlap %','Lang Detected']].to_string(index=False))
    print(f"\n  Avg chrF    : {df['chrF Score'].mean():.2f} / 100")
    print(f"  Avg ROUGE-L : {df['ROUGE-L'].mean():.3f} / 1.0")
    print(f"  Avg Overlap : {df['Word Overlap %'].mean():.1f}%")
    print(f"  Lang Check  : {(df['Lang Detected'] == 'en').sum()}/{len(df)} ✅")
    return df

df_tfidf_tamil     = evaluate_tfidf(tfidf_summary_clean, tfidf_tamil_summary,     'ta', 'Tamil')
df_tfidf_malayalam = evaluate_tfidf(tfidf_summary_clean, tfidf_malayalam_summary, 'ml', 'Malayalam')

# Side by side
print(f"\n{'='*65}")
print("  TF-IDF — TAMIL vs MALAYALAM COMPARISON")
print(f"{'='*65}")
print(f"  {'Metric':<22} {'Tamil':>10} {'Malayalam':>12}")
print(f"  {'─'*45}")
print(f"  {'Avg chrF':<22} {df_tfidf_tamil['chrF Score'].mean():>10.2f} {df_tfidf_malayalam['chrF Score'].mean():>12.2f}")
print(f"  {'Avg ROUGE-L':<22} {df_tfidf_tamil['ROUGE-L'].mean():>10.3f} {df_tfidf_malayalam['ROUGE-L'].mean():>12.3f}")
print(f"  {'Avg Overlap %':<22} {df_tfidf_tamil['Word Overlap %'].mean():>10.1f} {df_tfidf_malayalam['Word Overlap %'].mean():>12.1f}")
print(f"{'='*65}")
```

```python
# COMPLETE SETUP — after restart
import torch, sys, glob, os
sys.path.insert(0, '/content/IndicTrans2/huggingface_interface/IndicTransToolkit')
from transformers import AutoModelForSeq2SeqLM, AutoTokenizer
from IndicTransToolkit.processor import IndicProcessor

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
BATCH_SIZE = 4
print(f"Device: {DEVICE}")
print(f"Transformers: {__import__('transformers').__version__}")

def clean_translation(text):
    text = text.replace('</s>', '').replace('<pad>', '').replace('<s>', '')
    return ' '.join(text.split()).strip()

def batch_translate(input_sentences, src_lang, tgt_lang, model, tokenizer, ip):
    translations = []
    for i in range(0, len(input_sentences), BATCH_SIZE):
        batch = input_sentences[i:i+BATCH_SIZE]
        batch = ip.preprocess_batch(batch, src_lang=src_lang, tgt_lang=tgt_lang)
        inputs = tokenizer(batch, truncation=True, padding="longest",
                          return_tensors="pt", return_attention_mask=True).to(DEVICE)
        with torch.no_grad():
            generated_tokens = model.generate(
                **inputs, use_cache=True, min_length=0,
                max_length=256, num_beams=5, num_return_sequences=1)
        generated_tokens = tokenizer.batch_decode(
            generated_tokens, skip_special_tokens=True,
            clean_up_tokenization_spaces=True)
        translations += ip.postprocess_batch(generated_tokens, lang=tgt_lang)
        del inputs
        torch.cuda.empty_cache()
    return [clean_translation(t) for t in translations]

# Patch config
hash_path = '/root/.cache/huggingface/modules/transformers_modules/ai4bharat/indictrans2_hyphen_en_hyphen_indic_hyphen_1B/10e65a9951a1e922cd109a95e8aba9357b62144b'
config_path = os.path.join(hash_path, 'configuration_indictrans.py')
with open(config_path, 'r') as f: lines = f.readlines()
cleaned, skip = [], False
for line in lines:
    if 'transformers.onnx' in line: continue
    if 'class IndicTransOnnxConfig' in line: skip = True
    if skip and line.startswith('class ') and 'IndicTransOnnxConfig' not in line: skip = False
    if not skip: cleaned.append(line)
with open(config_path, 'w') as f: f.writelines(cleaned)
print("✅ Patch 1 — config")

tokenizer_path = os.path.join(hash_path, 'tokenization_indictrans.py')
with open(tokenizer_path, 'r') as f: content = f.read()
if '_special_tokens_map_data' not in content:
    patch = '''
    @property
    def _special_tokens_map(self):
        return getattr(self, '_special_tokens_map_data', {})

    @_special_tokens_map.setter
    def _special_tokens_map(self, value):
        self._special_tokens_map_data = value

    @property
    def additional_special_tokens(self):
        return getattr(self, '_additional_special_tokens', [])

    @additional_special_tokens.setter
    def additional_special_tokens(self, value):
        self._additional_special_tokens = value if value is not None else []

'''
    content = content.replace('def get_vocab(self)', f'{patch}\n    def get_vocab(self)')
    with open(tokenizer_path, 'w') as f: f.write(content)
print("✅ Patch 2 — tokenizer")

# Load tokenizer
print("Loading tokenizer...")
en_indic_tokenizer = AutoTokenizer.from_pretrained(
    "ai4bharat/indictrans2-en-indic-1B", trust_remote_code=True)
print("✅ Tokenizer loaded!")

# Patch modeling (now exists after tokenizer load)
all_modeling = glob.glob('/root/.cache/**/*modeling_indictrans*', recursive=True)
for path in all_modeling:
    with open(path, 'r') as f: content = f.read()
    content = content.replace('def tie_weights(self):', 'def tie_weights(self, **kwargs):')
    with open(path, 'w') as f: f.write(content)
    print(f"✅ Patch 3 — modeling")

# Load model
print("Loading model...")
en_indic_model = AutoModelForSeq2SeqLM.from_pretrained(
    "ai4bharat/indictrans2-en-indic-1B", trust_remote_code=True, low_cpu_mem_usage=True)
en_indic_model = en_indic_model.to(DEVICE)
en_indic_model.half()
en_indic_model.eval()
ip = IndicProcessor(inference=True)
print("✅ Model loaded!")

# Test both languages
result_tam = batch_translate(
    ["Customers are often disappointed with the value for money."],
    "eng_Latn", "tam_Taml", en_indic_model, en_indic_tokenizer, ip)
print(f"\nTamil test : {result_tam}")

result_mal = batch_translate(
    ["Customers are often disappointed with the value for money."],
    "eng_Latn", "mal_Mlym", en_indic_model, en_indic_tokenizer, ip)
print(f"Malayalam test: {result_mal}")
```

```python
# Evaluate TF-IDF IndicTrans2
from sacrebleu.metrics import CHRF
from rouge_score import rouge_scorer
from langdetect import detect
from deep_translator import GoogleTranslator
import pandas as pd

chrf   = CHRF()
scorer = rouge_scorer.RougeScorer(['rougeL'], use_stemmer=True)

def back_translate(text, src):
    try:
        return GoogleTranslator(source=src, target='en').translate(text)
    except:
        return ""

def evaluate(orig_texts, translations, src_lang, lang_name, feat_names):
    results = []
    for orig, trans, feat in zip(orig_texts, translations, feat_names):
        back = back_translate(trans, src_lang)
        if not back.strip(): continue
        orig_words = set(orig.lower().split())
        back_words = set(back.lower().split())
        results.append({
            'Feature'        : feat,
            'chrF Score'     : round(chrf.sentence_score(back, [orig]).score, 2),
            'ROUGE-L'        : round(scorer.score(orig, back)['rougeL'].fmeasure, 3),
            'Word Overlap %' : round(len(orig_words & back_words) / len(orig_words) * 100, 1),
            'Lang Detected'  : detect(back)
        })
    df = pd.DataFrame(results)
    print(f"\n{'='*65}\n  {lang_name.upper()} — EVALUATION\n{'='*65}")
    print(df[['Feature','chrF Score','ROUGE-L','Word Overlap %','Lang Detected']].to_string(index=False))
    print(f"\n  Avg chrF    : {df['chrF Score'].mean():.2f} / 100")
    print(f"  Avg ROUGE-L : {df['ROUGE-L'].mean():.3f} / 1.0")
    print(f"  Avg Overlap : {df['Word Overlap %'].mean():.1f}%")
    print(f"  Lang Check  : {(df['Lang Detected'] == 'en').sum()}/{len(df)} ✅")
    return df

df_tfidf_it2_tamil     = evaluate(tfidf_texts, tfidf_it2_tamil,
                                   'ta', 'TF-IDF IndicTrans2 Tamil',     tfidf_feat_names)
df_tfidf_it2_malayalam = evaluate(tfidf_texts, tfidf_it2_malayalam,
                                   'ml', 'TF-IDF IndicTrans2 Malayalam', tfidf_feat_names)

# GT baseline scores from earlier session
gt_tamil_avg     = {'chrF': 66.67, 'ROUGE-L': 0.730, 'Overlap': 70.0}
gt_malayalam_avg = {'chrF': 74.25, 'ROUGE-L': 0.774, 'Overlap': 76.2}

print(f"\n{'='*70}")
print("  TF-IDF: Google Translate vs IndicTrans2")
print(f"{'='*70}")
print(f"  {'Metric':<22} {'GT Tamil':>9} {'IT2 Tamil':>10} {'GT Malay':>10} {'IT2 Malay':>10}")
print(f"  {'─'*62}")
print(f"  {'Avg chrF':<22} {gt_tamil_avg['chrF']:>9.2f} {df_tfidf_it2_tamil['chrF Score'].mean():>10.2f} {gt_malayalam_avg['chrF']:>10.2f} {df_tfidf_it2_malayalam['chrF Score'].mean():>10.2f}")
print(f"  {'Avg ROUGE-L':<22} {gt_tamil_avg['ROUGE-L']:>9.3f} {df_tfidf_it2_tamil['ROUGE-L'].mean():>10.3f} {gt_malayalam_avg['ROUGE-L']:>9.3f} {df_tfidf_it2_malayalam['ROUGE-L'].mean():>10.3f}")
print(f"  {'Avg Overlap %':<22} {gt_tamil_avg['Overlap']:>9.1f} {df_tfidf_it2_tamil['Word Overlap %'].mean():>10.1f} {gt_malayalam_avg['Overlap']:>9.1f} {df_tfidf_it2_malayalam['Word Overlap %'].mean():>10.1f}")
print(f"{'='*70}")
print(f"\n  Winner Tamil   chrF   : {'IndicTrans2 ✅' if df_tfidf_it2_tamil['chrF Score'].mean() > gt_tamil_avg['chrF'] else 'Google Translate ✅'}")
print(f"  Winner Tamil   ROUGE-L: {'IndicTrans2 ✅' if df_tfidf_it2_tamil['ROUGE-L'].mean() > gt_tamil_avg['ROUGE-L'] else 'Google Translate ✅'}")
print(f"  Winner Malay   chrF   : {'IndicTrans2 ✅' if df_tfidf_it2_malayalam['chrF Score'].mean() > gt_malayalam_avg['chrF'] else 'Google Translate ✅'}")
print(f"  Winner Malay   ROUGE-L: {'IndicTrans2 ✅' if df_tfidf_it2_malayalam['ROUGE-L'].mean() > gt_malayalam_avg['ROUGE-L'] else 'Google Translate ✅'}")
print(f"{'='*70}")
```
