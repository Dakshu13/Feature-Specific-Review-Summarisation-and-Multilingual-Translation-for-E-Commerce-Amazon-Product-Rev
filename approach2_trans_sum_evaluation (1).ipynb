````python
from deep_translator import GoogleTranslator

def translate_summary(summary, target_lang):
    translated = {}

    # Translate Overall Summary
    overall = summary['Overall Summary']
    translated['Overall Summary'] = {
        'Overall Sentiment': GoogleTranslator(source='en', target=target_lang).translate(overall['Overall Sentiment']),
        'Summary':           GoogleTranslator(source='en', target=target_lang).translate(overall['Summary']),
        'Key Themes':        GoogleTranslator(source='en', target=target_lang).translate(overall['Key Themes'])
    }

    # Translate Feature-Based Summary
    translated['Feature-Based Summary'] = {}
    for feature, details in summary['Feature-Based Summary'].items():
        translated_feature = GoogleTranslator(source='en', target=target_lang).translate(feature)
        translated['Feature-Based Summary'][translated_feature] = {
            'Sentiment':   GoogleTranslator(source='en', target=target_lang).translate(details['Sentiment']),
            'Summary':     GoogleTranslator(source='en', target=target_lang).translate(details['Summary']),
            'Key Phrases': GoogleTranslator(source='en', target=target_lang).translate(details['Key Phrases'])
        }

    return translated

# Translate to Tamil and Malayalam
tamil_summary     = translate_summary(final_summary, 'ta')
malayalam_summary = translate_summary(final_summary, 'ml')

# Print Tamil
print("=" * 60)
print("        சுருக்கம் — தமிழ் (TAMIL)")
print("=" * 60)
overall = tamil_summary['Overall Summary']
print(f"உணர்வு    : {overall['Overall Sentiment']}")
print(f"முக்கிய தீம்: {overall['Key Themes']}")
print(f"\n{overall['Summary']}")

print("\n" + "=" * 60)
for feature, details in tamil_summary['Feature-Based Summary'].items():
    print(f"\n📌 {feature}")
    print(f"   உணர்வு  : {details['Sentiment']}")
    print(f"   சுருக்கம்: {details['Summary']}")
    print(f"   முக்கிய சொற்கள்: {details['Key Phrases']}")

# Print Malayalam
print("\n" + "=" * 60)
print("        സംഗ്രഹം — മലയാളം (MALAYALAM)")
print("=" * 60)
overall = malayalam_summary['Overall Summary']
print(f"വികാരം    : {overall['Overall Sentiment']}")
print(f"പ്രധാന തീമുകൾ: {overall['Key Themes']}")
print(f"\n{overall['Summary']}")

print("\n" + "=" * 60)
for feature, details in malayalam_summary['Feature-Based Summary'].items():
    print(f"\n📌 {feature}")
    print(f"   വികാരം  : {details['Sentiment']}")
    print(f"   സംഗ്രഹം : {details['Summary']}")
    print(f"   പ്രധാന വാക്കുകൾ: {details['Key Phrases']}")
````

````python
from sacrebleu.metrics import BLEU, CHRF
from rouge_score import rouge_scorer
from langdetect import detect
from deep_translator import GoogleTranslator
import pandas as pd

chrf = CHRF()
scorer = rouge_scorer.RougeScorer(['rougeL'], use_stemmer=True)

def back_translate(text, source_lang):
    try:
        return GoogleTranslator(source=source_lang, target='en').translate(text)
    except:
        return ""

def evaluate_translation(final_summary, translated_summary, source_lang, lang_name):
    originals, back_translated, features = [], [], []

    # Overall Summary
    orig = final_summary['Overall Summary']['Summary']
    trans_text = translated_summary['Overall Summary']['Summary']
    back = back_translate(trans_text, source_lang)
    originals.append(orig)
    back_translated.append(back)
    features.append('Overall Summary')

    # Feature summaries
    for feature, details in final_summary['Feature-Based Summary'].items():
        orig = details['Summary']
        if not orig.strip():
            continue
        idx = list(final_summary['Feature-Based Summary'].keys()).index(feature)
        trans_text = list(translated_summary['Feature-Based Summary'].values())[idx]['Summary']
        back = back_translate(trans_text, source_lang)
        originals.append(orig)
        back_translated.append(back)
        features.append(feature)

    results = []
    for orig, back, feat in zip(originals, back_translated, features):
        if not back.strip():
            continue
        chrf_score = chrf.sentence_score(back, [orig]).score
        rouge = scorer.score(orig, back)['rougeL'].fmeasure
        try:
            detected = detect(back)
        except:
            detected = "unknown"
        orig_words = set(orig.lower().split())
        back_words = set(back.lower().split())
        overlap = len(orig_words & back_words) / len(orig_words) * 100

        results.append({
            'Feature'        : feat,
            'chrF Score'     : round(chrf_score, 2),
            'ROUGE-L'        : round(rouge, 3),
            'Word Overlap %' : round(overlap, 1),
            'Lang Detected'  : detected
        })

    df = pd.DataFrame(results)

    print("=" * 70)
    print(f"        {lang_name.upper()} TRANSLATION EVALUATION METRICS")
    print("=" * 70)
    print(df[['Feature', 'chrF Score', 'ROUGE-L', 'Word Overlap %', 'Lang Detected']].to_string(index=False))
    print("\n" + "=" * 70)
    print("SUMMARY SCORES")
    print("=" * 70)
    print(f"  Avg chrF Score   : {df['chrF Score'].mean():.2f} / 100  (>40 = good)")
    print(f"  Avg ROUGE-L      : {df['ROUGE-L'].mean():.3f} / 1.0   (>0.4 = good)")
    print(f"  Avg Word Overlap : {df['Word Overlap %'].mean():.1f}%         (>30% = good)")
    print(f"  Correct Language : {(df['Lang Detected'] == 'en').sum()}/{len(df)} back-translations detected as English")
    print("=" * 70)

    return df

# ── Evaluate Tamil ──────────────────────────────────────────
df_tamil    = evaluate_translation(final_summary, tamil_summary,    'ta', 'Tamil')

print("\n\n")

# ── Evaluate Malayalam ──────────────────────────────────────
df_malayalam = evaluate_translation(final_summary, malayalam_summary, 'ml', 'Malayalam')

print("\n\n")

# ── Side-by-Side Comparison ─────────────────────────────────
print("=" * 70)
print("        TAMIL vs MALAYALAM — SIDE BY SIDE COMPARISON")
print("=" * 70)
comparison = pd.DataFrame({
    'Feature'          : df_tamil['Feature'],
    'Tamil chrF'       : df_tamil['chrF Score'],
    'Malayalam chrF'   : df_malayalam['chrF Score'],
    'Tamil ROUGE-L'    : df_tamil['ROUGE-L'],
    'Malayalam ROUGE-L': df_malayalam['ROUGE-L'],
    'Tamil Overlap%'   : df_tamil['Word Overlap %'],
    'Malayalam Overlap%': df_malayalam['Word Overlap %'],
})
print(comparison.to_string(index=False))
print("\n" + "=" * 70)
print("AVERAGE COMPARISON")
print("=" * 70)
print(f"  {'Metric':<20} {'Tamil':>10} {'Malayalam':>12}")
print(f"  {'─'*42}")
print(f"  {'chrF Score':<20} {df_tamil['chrF Score'].mean():>10.2f} {df_malayalam['chrF Score'].mean():>12.2f}")
print(f"  {'ROUGE-L':<20} {df_tamil['ROUGE-L'].mean():>10.3f} {df_malayalam['ROUGE-L'].mean():>12.3f}")
print(f"  {'Word Overlap %':<20} {df_tamil['Word Overlap %'].mean():>10.1f} {df_malayalam['Word Overlap %'].mean():>12.1f}")
print("=" * 70)
````

````python
# CELL 36: Load model + translate LLaMA summary
import torch
from transformers import AutoModelForSeq2SeqLM, AutoTokenizer
from IndicTransToolkit.processor import IndicProcessor

BATCH_SIZE = 4
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Device: {DEVICE}")

def initialize_model_and_tokenizer(ckpt_dir):
    tokenizer = AutoTokenizer.from_pretrained(ckpt_dir, trust_remote_code=True)
    model = AutoModelForSeq2SeqLM.from_pretrained(
        ckpt_dir, trust_remote_code=True, low_cpu_mem_usage=True)
    if DEVICE == "cuda":
        model = model.to(DEVICE)
        model.half()
    model.eval()
    return tokenizer, model

def batch_translate(input_sentences, src_lang, tgt_lang, model, tokenizer, ip):
    translations = []
    for i in range(0, len(input_sentences), BATCH_SIZE):
        batch = input_sentences[i:i+BATCH_SIZE]
        batch = ip.preprocess_batch(batch, src_lang=src_lang, tgt_lang=tgt_lang)
        inputs = tokenizer(batch, truncation=True, padding="longest",
                          return_tensors="pt", return_attention_mask=True).to(DEVICE)
        with torch.no_grad():
            generated_tokens = model.generate(
                **inputs, use_cache=True, min_length=0,
                max_length=256, num_beams=5, num_return_sequences=1)
        generated_tokens = tokenizer.batch_decode(
            generated_tokens, skip_special_tokens=True,
            clean_up_tokenization_spaces=True)
        translations += ip.postprocess_batch(generated_tokens, lang=tgt_lang)
        del inputs
        torch.cuda.empty_cache()
    return translations

def clean_translation(text):
    text = text.replace('</s>', '').replace('<pad>', '').replace('<s>', '')
    text = ' '.join(text.split())
    return text.strip()

print("Loading model...")
en_indic_tokenizer, en_indic_model = initialize_model_and_tokenizer(
    "ai4bharat/indictrans2-en-indic-1B")
ip = IndicProcessor(inference=True)
print("✅ Model loaded!")

# Texts from final_summary
it2_texts = [
    final_summary['Overall Summary']['Summary'],
    final_summary['Overall Summary']['Key Themes'],
] + [v['Summary'] for v in final_summary['Feature-Based Summary'].values()]

it2_feature_names = ['Overall Summary', 'Key Themes', 'Product Quality', 'Value for Money',
                     'Packaging & Presentation', 'Delivery & Shipping',
                     'Variety & Curation', 'Customer Service']

print("\nTranslating to Tamil...")
it2_tamil = [clean_translation(t) for t in
             batch_translate(it2_texts, "eng_Latn", "tam_Taml",
                             en_indic_model, en_indic_tokenizer, ip)]
print("✅ Tamil done!")

print("Translating to Malayalam...")
it2_malayalam = [clean_translation(t) for t in
                 batch_translate(it2_texts, "eng_Latn", "mal_Mlym",
                                 en_indic_model, en_indic_tokenizer, ip)]
print("✅ Malayalam done!")

print("\n" + "="*60 + "\n  LLAMA — INDICTRANS2 TAMIL\n" + "="*60)
for feat, trans in zip(it2_feature_names, it2_tamil):
    print(f"\n📌 {feat}\n   {trans}")

print("\n" + "="*60 + "\n  LLAMA — INDICTRANS2 MALAYALAM\n" + "="*60)
for feat, trans in zip(it2_feature_names, it2_malayalam):
    print(f"\n📌 {feat}\n   {trans}")
````

````python
# CELL 37: IndicTrans2 Evaluation
!pip install sacrebleu rouge-score langdetect deep-translator --quiet

from sacrebleu.metrics import CHRF
from rouge_score import rouge_scorer
from langdetect import detect
from deep_translator import GoogleTranslator
import pandas as pd

chrf   = CHRF()
scorer = rouge_scorer.RougeScorer(['rougeL'], use_stemmer=True)

def back_translate(text, src):
    try:
        return GoogleTranslator(source=src, target='en').translate(text)
    except:
        return ""

def evaluate(orig_texts, translations, src_lang, lang_name, feat_names):
    results = []
    for orig, trans, feat in zip(orig_texts, translations, feat_names):
        back = back_translate(trans, src_lang)
        if not back.strip(): continue
        orig_words = set(orig.lower().split())
        back_words = set(back.lower().split())
        results.append({
            'Feature'        : feat,
            'chrF Score'     : round(chrf.sentence_score(back, [orig]).score, 2),
            'ROUGE-L'        : round(scorer.score(orig, back)['rougeL'].fmeasure, 3),
            'Word Overlap %' : round(len(orig_words & back_words) / len(orig_words) * 100, 1),
            'Lang Detected'  : detect(back)
        })
    df = pd.DataFrame(results)
    print(f"\n{'='*65}\n  {lang_name.upper()} EVALUATION\n{'='*65}")
    print(df[['Feature','chrF Score','ROUGE-L','Word Overlap %','Lang Detected']].to_string(index=False))
    print(f"\n  Avg chrF    : {df['chrF Score'].mean():.2f} / 100")
    print(f"  Avg ROUGE-L : {df['ROUGE-L'].mean():.3f} / 1.0")
    print(f"  Avg Overlap : {df['Word Overlap %'].mean():.1f}%")
    print(f"  Lang Check  : {(df['Lang Detected'] == 'en').sum()}/{len(df)} ✅")
    return df

# Evaluate IndicTrans2 Tamil and Malayalam
df_it2_tamil     = evaluate(it2_texts, it2_tamil,     'ta', 'IndicTrans2 — Tamil',     it2_feature_names)
df_it2_malayalam = evaluate(it2_texts, it2_malayalam, 'ml', 'IndicTrans2 — Malayalam', it2_feature_names)
````
